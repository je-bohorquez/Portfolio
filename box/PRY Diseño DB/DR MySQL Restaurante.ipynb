-- Usar la base de datos Restaurante
USE Restaurante;

-- Tabla Restaurante 01
CREATE TABLE Restaurante (
    ID_Restaurante CHAR(10),
    ID_Carta CHAR(10),
    Id_Pedido CHAR(10),
    Nombre CHAR(10),
    Ubicacion NVARCHAR(30),
    PRIMARY KEY (ID_Restaurante)
);

-- Tabla Carta 02
CREATE TABLE Carta (
    ID_Carta CHAR(10) PRIMARY KEY,
    ID_Restaurante CHAR(10),
    ID_Plato CHAR(10),
    Descripcion CHAR(10)
);

-- Tabla Receta 03
CREATE TABLE Receta (
    ID_Receta CHAR(10) PRIMARY KEY,
    ID_Ingrediente CHAR(10),
    ID_Medida CHAR(10),
    Descripcion NVARCHAR(50),
    Cantidad INT
);

-- Tabla Ingredientes 04
CREATE TABLE Ingredientes (
    ID_Ingredientes CHAR(10) PRIMARY KEY,
    Nombre CHAR(10),
    PrecioCompra CHAR(10),
    StockAlmacen INT
);

-- Tabla Plato 05
CREATE TABLE Plato (
    ID_Plato CHAR(10) PRIMARY KEY,
    ID_Dificultad CHAR(10),
    ID_Categoria CHAR(10),
    Nombre CHAR(10),
    Descripcion NVARCHAR(50),
    Precio DECIMAL(18, 0)
);

-- Tabla NDificultad 06
CREATE TABLE NDificultad (
    ID_NDificultad CHAR(10) PRIMARY KEY,
    Descripcion CHAR(15)
);

-- Tabla Categoria 07
CREATE TABLE Categoria (
    ID_Categoria CHAR(10) PRIMARY KEY,
    Descripcion CHAR(10),
    ID_Encargado CHAR(10)
);

-- Tabla Encargado 08
CREATE TABLE Encargado (
    ID_Encargado CHAR(10) PRIMARY KEY,
    Nombre CHAR(10),
    Apellidos CHAR(10),
    Edad INT,
    DNI NUMERIC(18, 0),
    Direccion CHAR(30)
);

-- Tabla Medida 09
CREATE TABLE Medida (
    ID_Medida CHAR(10) PRIMARY KEY,
    Direccion CHAR(20)
);

-- Tabla PedidoVenta 10
CREATE TABLE PedidoVenta (
    ID_Pedido CHAR(10) PRIMARY KEY,
    ID_DetallePedido CHAR(10),
    Fecha DATE,
    MontoFinal DECIMAL(18, 0)
);

-- Tabla DetallePedido 11
CREATE TABLE DetallePedido (
    ID_Pedido CHAR(10) PRIMARY KEY,
    ID_Plato CHAR(10),
    Subtotal DECIMAL(18, 0),
    Cantidad INT
);

ALTER TABLE Carta
ADD FOREIGN KEY (ID_Plato) REFERENCES Plato(ID_Plato);

-- Tabla Receta (Relación con Ingredientes)
ALTER TABLE Receta
ADD FOREIGN KEY (ID_Ingrediente) REFERENCES Ingredientes(ID_Ingredientes);

-- Tabla Plato (Relaciones con NDificultad y Categoria)
ALTER TABLE Plato
ADD FOREIGN KEY (ID_Dificultad) REFERENCES NDificultad(ID_NDificultad);
ALTER TABLE Plato
ADD FOREIGN KEY (ID_Categoria) REFERENCES Categoria(ID_Categoria);

-- Tabla Categoria (Relación con Encargado)
ALTER TABLE Categoria
ADD FOREIGN KEY (ID_Encargado) REFERENCES Encargado(ID_Encargado);

-- Tabla PedidoVenta (Relación con DetallePedido)
ALTER TABLE PedidoVenta
ADD FOREIGN KEY (ID_DetallePedido) REFERENCES DetallePedido(ID_Pedido);
